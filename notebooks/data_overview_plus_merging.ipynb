```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA_DIR = Path("../data")

main_path = DATA_DIR / "DataSheet_Main.csv"
country_meta_path = DATA_DIR / "Metadata_Country.csv"
indicator_meta_path = DATA_DIR / "Metadata_Indicators.csv"

gdp_path = DATA_DIR / "DataSheet_GDP.csv"
country_meta_gdp_path = DATA_DIR / "Metadata_Country_GDP.csv"
indicator_meta_gdp_path = DATA_DIR / "Metadata_Indicator_GDP.csv"
```

```python
df = pd.read_csv(main_path, skiprows = 4)
country_meta = pd.read_csv(country_meta_path)
indicator_meta = pd.read_csv(indicator_meta_path)

df_gdp = pd.read_csv(gdp_path, skiprows = 4)
country_meta_gdp = pd.read_csv(country_meta_gdp_path)
indicator_meta_gdp = pd.read_csv(indicator_meta_gdp_path)

df = df.dropna(axis = 1, how="all")
country_meta = country_meta.dropna(axis = 1, how = "all")
indicator_meta = indicator_meta.dropna(axis = 1, how = "all")

df_gdp = df_gdp.dropna(axis = 1, how="all")
country_meta_gdp = country_meta_gdp.dropna(axis = 1, how = "all")
indicator_meta_gdp = indicator_meta_gdp.dropna(axis = 1, how = "all")
```

```python
print("Main dataset:")
display(df)
print(df.shape)
print(df.columns)

print("Main dataset:")
display(df_gdp)
print(df_gdp.shape)
print(df_gdp.columns)
```

```python
with pd.option_context("display.max_rows", None):
    display(indicator_meta)

with pd.option_context("display.max_rows", None):
    display(indicator_meta_gdp)
```

```python
selected_codes = [
    "NY.GDP.MKTP.KD.ZG",
    "FS.AST.PRVT.GD.ZS",
    "FB.CBK.BRCH.P5",
    "FB.ATM.TOTL.P5",
    "CM.MKT.LCAP.GD.ZS",
    "FR.INR.LNDP",
    "FM.LBL.BMNY.GD.ZS",
    "BX.KLT.DINV.WD.GD.ZS",
    "FB.AST.NPER.ZS",
    "NY.GDP.DEFL.KD.ZG.AD"
]

df_filtered = df[df["Indicator Code"].isin(selected_codes)]
display(df_filtered)
```

```python
df_gdp_filtered = df_gdp[df_gdp["Indicator Code"].isin(selected_codes)]

display(df_gdp_filtered)
```

```python
display(country_meta)
```

```python
regions = country_meta[
    country_meta["Region"].isna() & country_meta["IncomeGroup"].isna()
]

region_codes = regions["Country Code"].tolist()

#display(regions)

is_region = df_filtered["Country Code"].isin(region_codes)
df_filtered_no_regions = df_filtered[is_region == False]
#df_filtered_just_regions = df_filtered[is_region == True]

display(df_filtered_no_regions)
#display(df_filtered_just_regions)
#display(df_filtered)
```

```python
regions_gdp = country_meta_gdp[
    country_meta_gdp["Region"].isna() & country_meta_gdp["IncomeGroup"].isna()
]

region_codes_gdp = regions_gdp["Country Code"].tolist()

is_region_gdp = df_gdp_filtered["Country Code"].isin(region_codes_gdp)
df_gdp_filtered_no_regions = df_gdp_filtered[is_region_gdp == False]

display(df_gdp_filtered_no_regions)
```

```python
year_cols = [str(year) for year in range(1960, 2024)]
werte_pro_jahr = df_filtered_no_regions[year_cols].notna().sum()

display(werte_pro_jahr)
```

```python
year_cols_gdp = [str(year) for year in range(1961, 2024)]
werte_pro_jahr_gdp = df_gdp_filtered_no_regions[year_cols_gdp].notna().sum()
display(werte_pro_jahr_gdp)
```

```python
years_to_keep = [str(year) for year in range(2004, 2022)]
id_cols = ["Country Name", "Indicator Name"]
df_filtered_no_regions_relevant_years = df_filtered_no_regions[id_cols + years_to_keep]
display(df_filtered_no_regions_relevant_years)
```

```python
df_gdp_filtered_no_regions_relevant_years = df_gdp_filtered_no_regions[id_cols + years_to_keep]

display(df_gdp_filtered_no_regions_relevant_years)
```

```python
df_filtered_no_regions_relevant_years = pd.concat(
    [
        df_filtered_no_regions_relevant_years,
        df_gdp_filtered_no_regions_relevant_years
    ],
    ignore_index=True
)

display(df_filtered_no_regions_relevant_years)
```

```python
print(df_filtered_no_regions_relevant_years.shape)
print(df_filtered_no_regions_relevant_years.columns)
```

```python
df_long = df_filtered_no_regions_relevant_years.melt(
    id_vars = id_cols,
    value_vars = years_to_keep,
    var_name = "Year",
    value_name = "Value"
)

df_long["Year"] = df_long["Year"].astype(int)
df_long["Value"] = pd.to_numeric(df_long["Value"], errors="coerce")

display(df_long)
print(df_long.shape)
```

```python
df_panel = df_long.pivot_table(
    index = ["Country Name", "Year"],
    columns = ["Indicator Name"],
    values = "Value"
).reset_index()

display(df_panel)
print(df_panel.shape)
print(df_panel.columns)
```

```python
df_panel = df_panel.rename(columns={
    "GDP growth (annual %)": "gdp_growth",
    "Domestic credit to private sector (% of GDP)": "domestic_credit",
    "Commercial bank branches (per 100,000 adults)": "bank_branches",
    "Automated teller machines (ATMs) (per 100,000 adults)": "atms",
    "Market capitalization of listed domestic companies (% of GDP)": "market_cap",
    "Interest rate spread (lending rate minus deposit rate, %)": "interest_spread",
    "Broad money (% of GDP)": "broad_money",
    "Foreign direct investment, net inflows (% of GDP)": "fdi_inflows",
    "Bank nonperforming loans to total gross loans (%)": "nonperforming_loans",
    "Inflation, GDP deflator: linked series (annual %)": "inflation"
})

display(df_panel)
```
